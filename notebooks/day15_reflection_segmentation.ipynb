## 1. Observations, Interpretation, and Discussion

**Observations:** 
The quantitative evaluation using Absolute Count Error (ACE) shows that the Baseline configuration achieved an ACE of 0, correctly identifying all 30 beans. A lower threshold ($T_{Otsu}-0.05$) increased ACE due to under-segmentation, which drastically raised both the mean area and its standard deviation. Conversely, reducing the morphological radius (r=1) caused over-segmentation, sharply increasing ACE and the standard deviation of eccentricity.

**Interpretation:** 
The Otsu threshold effectively separates the dark coffee beans from the bright background. However, lowering the threshold incorporates inter-bean shadows into the foreground mask, merging adjacent beans into single, oversized connected components (increasing area variance). Smaller morphological kernels (r=1) fail to fully close the natural central crease found in coffee beans, causing individual beans to split into multiple small fragments, which skews the count upward and introduces irregular, highly eccentric shapes.

**Discussion and Critical Reflection:** 
The Baseline configuration ($T_{Otsu}$, r=3) is highly suitable for our automated quality inspection use case, accurately satisfying the core objective of reliable bean counting. However, the system's high sensitivity to minor threshold shifts reveals a practical limitation regarding uneven lighting. The ACE metric, combined with the standard deviation of object area, effectively captures overall segmentation performance; because counting accuracy and size uniformity serve as excellent direct proxies for correct object separation when ground-truth pixel masks are unavailable.

## 2. Personal Final Reflection on MC

What worked particularly well was the interplay between morphological operations and thresholding. I was impressed by how a simple morphological closing operation (r=3) robustly resolved the internal holes caused by the central bean crease, immediately transforming fragmented pixel clusters into coherent, measurable objects. 

If I repeated the task, I would approach the evaluation phase differently by establishing a quantitative baseline earlier. Initially, relying solely on visual inspection made parameter tuning somewhat subjective. An insightful and unexpected observation was how drastically a very small shift in the threshold ($0.05$) degraded segmentation performance by merging adjacent beans through their faint cast shadows, revealing that even seemingly 'clean' image setups contain hidden complexities.

This experience highlighted a significant gap between theoretical methods and practical application. Algorithms like Otsu thresholding theoretically assume perfectly bimodal histograms, but real-world textures and shadows blur these boundaries. Furthermore, theoretical morphology often assumes idealized geometric shapes, yet the natural biological variation in the beans (creases, irregular edges) required empirical kernel tuning rather than strict mathematical derivation. It reinforced that robust applied image processing heavily depends on deeply understanding the specific physical characteristics of the domain data.
